```python
# More specified objective function visualizer

w=1
A=6.7
p=0.1

q1=1
p1=0
q2=2
p2=1
q3=4
p3=-3

c1=p1/q1
x1 = np.linspace(c1-p/2,c1+p/2,50)
x1_1 = np.linspace(-7,c1-p/2,2)
x1_2 = np.linspace(c1+p/2,7,2)

c2=p2/q2
x2 = np.linspace(c2-p/2,c2+p/2,50)
x2_1 = np.linspace(-7,c2-p/2,2)
x2_2 = np.linspace(c2+p/2,7,2)

c3=p3/q3
x3 = np.linspace(c3-p/2,c3+p/2,50)
x3_1 = np.linspace(-7,c3-p/2,2)
x3_2 = np.linspace(c3+p/2,7,2)


plt.figure(1)
a1 = c1-p/2
b1 = c1+p/2
a2 = c2-p/2
b2 = c2+p/2
a3 = c3-p/2
b3 = c3+p/2
y1 = A*np.exp( w * ((a1-b1)**2) / ( (x1-a1)*(x1-b1) ) ) / q1
y2 = A*np.exp( w * ((a1-b1)**2) / ( (x2-a2)*(x2-b2) ) ) / q1
y3 = A*np.exp( w * ((a1-b1)**2) / ( (x3-a3)*(x3-b3) ) ) / q3

# plt.plot(x3,y3,"r",label=r"$\omega_{\zeta}^{res}$ = "+str(c1))
# plt.plot(x3_1,np.array([0,0]),"r")
# plt.plot(x3_2,np.array([0,0]),"r")
fig, ax = plt.subplots(2,1,figsize=(8,6))
# plt.plot(x1,y1,"b",label=r"$\omega_{\zeta}^{res}$ = "+str(c2))
ax[0].plot(x1,y1,"b")
ax[0].plot(x1_1,np.array([0,0]),"b")
ax[0].plot(x1_2,np.array([0,0]),"b")
# plt.plot(x2,y2,"g")
# plt.plot(x2_1,np.array([0,0]),"g")
# plt.plot(x2_2,np.array([0,0]),"g")
ax[0].plot(-x2,y2,"purple")
ax[0].plot(-x2_1,np.array([0,0]),"purple")
ax[0].plot(-x2_2,np.array([0,0]),"purple")
# plt.plot(1+x1,y1,"b")
# plt.plot(1+x1_1,np.array([0,0]),"b")
# plt.plot(1+x1_2,np.array([0,0]),"b")
ax[0].plot(-1+x1,y1,"darkgreen")
ax[0].plot(-1+x1_1,np.array([0,0]),"darkgreen")
ax[0].plot(-1+x1_2,np.array([0,0]),"darkgreen")

# plt.title("Resonance Objective Function Terms")
# ax[0].set_xlabel(r"$\Omega_{\eta}$",fontsize=20)
ax[0].set_ylabel(r"$f_b$",fontsize=27)
ax[0].set_xlim([-1.1,0.15])
ax[0].tick_params(axis='x', labelsize=16)
# ax[2].tick_params(axis='y', labelsize=16)
# ax.yaxis.set_visible(False)
# ax.set_yticklabels([])
# plt.legend()
labels = [item.get_text() for item in ax[0].get_yticklabels()]
labels[2:]=''
labels[1] = '0.0'
ax[0].set_yticklabels(labels,fontsize=16)

xs = [-1.0,-0.62,-0.37,-0.2,-0.1,-0.05,-0.001,0.034,0.05,0.099]
i=0
for x in xs:
    if i==0:
        ax[0].scatter(x,0,s=100,marker='x',color='black',linewidth=5,label=r'$\Omega_\eta(\rho)$')
    else:
        ax[0].scatter(x,0,s=100,marker='x',color='black',linewidth=5)
    i+=1
ax[0].legend(fontsize=20,loc="upper left",bbox_to_anchor=(0.5, 1))
# plt.tight_layout()
# plt.savefig("res_visual.pdf")

# fig1,ax1 = plt.subplots()
ys = np.linspace(0,1,len(xs))
ax[1].plot(xs,ys,'-xk',ms=10,mew=5)
ax[1].set_xlabel(r"$\Omega_{\eta}$",fontsize=20)
ax[1].set_ylabel(r"$\rho$",fontsize=20)
ax[1].tick_params(axis='x', labelsize=16)
ax[1].tick_params(axis='y', labelsize=16)
ax[1].set_xticks([-1.00,-0.75,-0.50,-0.25,0.00])
ax[0].set_xticks([-1.00,-0.75,-0.50,-0.25,0.00])
ax[0].set_xticklabels([])
ax[1].set_xlim([-1.1,0.15])
plt.tight_layout()
plt.savefig("res_visual_tg.pdf")
```
