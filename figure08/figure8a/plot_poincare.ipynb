```python
import matplotlib.pyplot as plt 
import numpy as np 
plt.rcParams['font.size'] = 14 # Sets the default font size (in points)
```

```python
data = np.load('_poincare_data_Bcrit_4.67.npz',allow_pickle=True)
s_all = data['s_all']
etas_all = data['etas_all']

num_traj = s_all.shape[0]

plt.figure()
for i in range(num_traj):
    eta = np.mod(etas_all[i,:],2*np.pi)
    plt.scatter(eta,s_all[i,:],marker="o",s=0.5,edgecolors="none",rasterized=True)

plt.ylabel(r'$s$')
plt.xlabel(r'$\eta$')
plt.xlim(0,2*np.pi)
plt.ylim(0,1)
plt.savefig('poincare_Bcrit_4.67.pdf')
plt.savefig('poincare_Bcrit_4.67.png',dpi=600)
```
